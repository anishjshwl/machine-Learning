# Task 3  Outlier and IQR

```python
sns.boxplot(y = mydata["Close"])
plt.title("Boxplot  of closing ")
plt.show()
```

```python
Q1 = mydata["Close"].quantile(0.25)  # first quartile 
Q3 = mydata["Close"].quantile(0.75) # third quartile
IQR = Q3-Q1
lower_limit = Q1-1.5*IQR
upper_limit = Q3 + 1.5 * IQR
print("Q1 = " , Q1)
print("Q3 = " , Q3)
print("IQR = ", IQR)
print("lower limit = " ,lower_limit)
print("upper limit = ",upper_limit)

```

#  Task 5 Closing Price Trend

```python
plt.Figure(figure = (20,50))
for Sector in mydata["Sector"].unique():
    temp =mydata[mydata["Sector"] == Sector]
    plt.plot(temp["Date"] , temp["Close"],label =Sector)

plt.title("closing Price Trend")
plt.xlabel("Date")
plt.ylabel("close")
plt.legend()
plt.show()
```

#  Task 6 Daily Return and Rolling Average

```python
mydata = mydata.sort_values(["Ticker", "Date"])
mydata["Return"] = mydata.groupby("Ticker")["Close"].pct_change()
result = mydata.groupby("NewsSentiment")["Close"].mean()
result

mydata["Rolling_Avg"] = (
    mydata.groupby("Ticker")["Close"]
          .transform(lambda x: x.rolling(window=7).mean())
)
mydata.head()
```

# Task 7 Sector-wise Averages and Volatility.

```python
mydata.groupby("Sector")["Close"].mean()
```

```python
mydata.groupby("Sector")["Close"].std()
```

# Task 8 Correlation Matrix and Heatmap

```python
cor_mat= mydata.corr(numeric_only = True)   # cor_mat  means correlation matrix
cor_mat

sns.heatmap(cor_mat)

sns.heatmap( cor_mat, 
            annot = True,
            cmap = "coolwarm",
            linewidth = 1.5
           )
plt.title("correlation matrix and heatmap")
plt.show()
```
