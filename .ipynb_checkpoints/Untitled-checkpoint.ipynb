```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns


data = pd.read_csv("ms_practice.csv")
x = [0,1,2,3,4,5]
y = [10,20,30,40,50,100]
plt.plot(x,y)  #   it is necessary for the  data set  
plt.xlabel("trial")  # name of the   x label
plt.ylabel("practice")  # name of the y label 
plt.title("trial of the data") # title of  the project 
plt.plot(x,y, color= "blue") # we can aslo use the  color
plt.plot(x,y,linestyle = "dotted" ) #  we have more option  for linesyle like( dashed, dotted , dash-dotted, solid) 
plt.plot(x,y, marker = "o")   # this means add Marker &  we have more option ("o", "s", *, x,+, ^)

plt.show() #  it show the data 

```

```python
fig, axis = plt.subplots(nrows  = 3, ncols =3 ,figsize = (10,10))   # fig represent the whole figure(canvas) , axis is making individual  graph
print(axis)
axis[0,0].plot([1,2,3],[4,5,6])
axis[1,1].plot([1,2,3],[6,5,4])
axis[2,2].plot([1,2,3],[2,4,8])
plt.show()
```

```python
fig , axis =plt.subplots(3,3,figsize = (10,10))
for row in range(3):
    for cols in range(3):
        axis[row,cols].plot([1,2,3], [4,5,6])
        print(row, cols)
```

```python
fig, axis = plt.subplots(figsize = (10,10))

```

```python
plot = sns.boxplot(x = "exteriorst", y = "saleprice", data = data,ax = axis);
plot.set_xticklabels(plot.get_xticklabels(), rotation = 90);
plt.show()
```
